```python
import os
import glob
from src.analyzer import IQSignalAnalyzer

# 1. Definimos la ruta base donde están todas las carpetas de los drones
ruta_base = "data/raw/"

# 2. Buscamos todos los archivos .iq de forma recursiva (el ** indica que busque en subcarpetas)
archivos_iq = glob.glob(os.path.join(ruta_base, "**", "*.iq"), recursive=True)

print(f"Se han encontrado un total de {len(archivos_iq)} archivos .iq en el sistema.")

# 3. Límite de seguridad para evitar el colapso de la memoria del navegador
# Para la reunión, te sugiero mostrar los primeros 4 o 5.
LIMITE_GRAFICAS = 21 

for indice, archivo in enumerate(archivos_iq):
    if indice >= LIMITE_GRAFICAS:
        print(f"\n--- PAUSA DE SEGURIDAD ---")
        print(f"Se ha alcanzado el límite de {LIMITE_GRAFICAS} gráficas para proteger la RAM de Jupyter.")
        print("Si deseas ver más, modifica la variable LIMITE_GRAFICAS.")
        break
        
    print(f"\n[{indice + 1}/{LIMITE_GRAFICAS}] Procesando archivo: {os.path.basename(archivo)}")
    
    try:
        # Instanciamos el analizador para el archivo actual
        analyzer = IQSignalAnalyzer(iq_file_path=archivo)
        
        # Leemos 50 milisegundos con un offset de 0.1s para evitar el ruido de inicialización del hardware SDR
        senal = analyzer.read_binary_chunk(duration=0.05, offset_sec=0)
        
        # Generamos la gráfica (ahora incluirá los puntos de máxima/mínima energía y la leyenda)
        analyzer.generate_analysis_panel(senal)
        
    except Exception as e:
        print(f"Error procesando el archivo {archivo}: {str(e)}")
```

```python
import numpy as np

# 1. Ponemos la ruta exacta de uno de tus ficheros
archivo_iq = "data/raw/DJI MINI3/DJI MINI3/VTSBW=10/pack2_0-1s.iq" 

# Queremos ver solo 10 muestras de la señal
num_muestras = 10

print("--- LEYENDO EL FICHERO EN CRUDO ---")
with open(archivo_iq, 'rb') as f:
    # Leemos 20 números flotantes (porque cada muestra tiene 2 partes: I y Q)
    raw_data = np.fromfile(f, dtype=np.float32, count=num_muestras * 2)

print("\n1. Vector Plano Original (Tal cual sale del disco duro):")
print(raw_data)
print(f"Tipo: {raw_data.dtype} | Cantidad de números: {len(raw_data)}")

print("\n--- SEPARANDO COMPONENTES (SLICING) ---")
# Cogemos los pares (índices 0, 2, 4...)
I = raw_data[0::2]
# Cogemos los impares (índices 1, 3, 5...)
Q = raw_data[1::2]

print(f"Componente I (En Fase):    {I}")
print(f"Componente Q (Cuadratura): {Q}")

print("\n--- RECONSTRUCCIÓN MATEMÁTICA ---")
# Creamos el número complejo final: I + jQ
complex_signal = I + 1j * Q

print("3. Vector Complejo Final (Lo que le pasamos a la STFT):")
print(complex_signal)
print(f"Tipo: {complex_signal.dtype} | Cantidad de muestras: {len(complex_signal)}")
```
